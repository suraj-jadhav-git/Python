```python
# 1
import pandas as pd
numbers = [10, 20, 30, 40, 50]
s = pd.Series(numbers)
print(s)
```

```python
# 2
marks = {
    "Rahul": 80,
    "Priya": 90,
    "Amit": 75
}

s = pd.Series(marks)

print(s)
```

```python
# 3
students = [
    {"Name": "Rahul", "Age": 20},
    {"Name": "Priya", "Age": 21},
    {"Name": "Amit", "Age": 22}
]

df = pd.DataFrame(students)

print(df)
```

```python
# 4
df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit"]
})

df["Country"] = "India"

print(df)
```

```python
# 5
df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit"]
})

ages = pd.Series([20, 21, 22])

df["Age"] = ages

print(df)
```

```python
# 6 one coln
df = pd.DataFrame({
    "Name": ["Rahul", "Priya"],
    "Age": [20, 21]
})

df = df.rename(columns={"Age": "Student_Age"})

print(df)




# rename multiple coln

df = pd.DataFrame({
    "Name": ["Rahul", "Priya"],
    "Age": [20, 21]
})

df = df.rename(columns={
    "Name": "Student_Name",
    "Age": "Student_Age"
})

print(df)
```

```python
# 7
df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit"],
    "Age": [20, 21, 22]
})

print(df)

# select single row
print("single row",df.loc[0])
# select multiple row
print("multiple rows",df.loc[[0, 2]])
# select conditional row
print("selective rows",df.loc[df["Age"] > 20])
```

```python
# 8

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit"],
    "Age": [20, 21, 22]
})

print("Shape:")
print(df.shape)

print("Data Types:")
print(df.dtypes)
```

```python
# 10
import numpy as np

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit"],
    "Marks": [80, np.nan, 75]
})

print("Before fillna():")
print(df)

df["Marks"] = df["Marks"].fillna(0)

print("\nAfter fillna():")
print(df)
```

Pandas Interview Questions (Theory Answers)

## 1. Difference between Series and DataFrame?

A Series is a one-dimensional data structure that stores a single column of data.

A DataFrame is a two-dimensional data structure that stores data in rows and columns, like an Excel table.

---

## 2. loc vs iloc?

loc is used to select data using row labels, index names, or conditions.

iloc is used to select data using row and column position numbers.

---

## 3. Why is Pandas built on NumPy?

Pandas is built on NumPy because NumPy provides fast and efficient array operations.

Using NumPy makes Pandas faster and more powerful for data analysis.

---

## 4. What is heterogeneous data?

Heterogeneous data means data that contains different data types together.

For example, a table may contain names (text), ages (numbers), and results (True/False).

---

## 5. How do you rename columns?

Columns can be renamed using the rename() function.

It allows us to change old column names to new column names.

---

## 6. How do you add columns?

A new column can be added by assigning values to a new column name.

Pandas automatically creates the column and stores the values.

---

## 7. How does Pandas handle missing values?

Pandas represents missing values using NaN (Not a Number).

Missing values can be handled using functions such as fillna() and dropna().

---

## 8. Explain DataFrame attributes.

DataFrame attributes provide information about the DataFrame.

Common attributes are:

* shape → number of rows and columns
* columns → column names
* dtypes → data types of columns
* index → row labels

---

## 9. Difference between List of Dictionaries and Dictionary of Series?

In a list of dictionaries, each dictionary represents a row.

In a dictionary of Series, each Series represents a column.

A list of dictionaries is row-oriented, while a dictionary of Series is column-oriented.

---

## 10. When would you choose loc over iloc?

Use loc when selecting data using labels, column names, or conditions.

Use iloc when selecting data using row and column position numbers.

loc is preferred when the actual row labels or column names are known.
