### Exercise Set 3 — Data Cleaning
```
7. Given a DataFrame with missing values, fill numeric NaN with column mean,
   and string NaN with 'Unknown'.
8. Find and remove duplicate rows.
9. Convert an 'Age' column stored as strings to integers.
10. Use str.strip() and str.title() to clean a messy 'Name' column.

````python
# 7
import numpy as np
data = {
    "Name": ["swetali", np.nan, "suraj", "pranu"],
    "Age": [25, np.nan, 30, 28],
    "Salary": [40000, 50000, np.nan, 60000],
    "Department": ["HR", "IT", np.nan, "DS"]
}

df = pd.DataFrame(data)
print("Original DataFrame")
print(df)
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df["Name"] = df["Name"].fillna("Unknown")
df["Department"] = df["Department"].fillna("Unknown")
print(df)
````

### Exercise Set 4 — Manipulation & GroupBy
```
11. Filter employees with Score > 80, sort by Score descending.
12. Use groupby to find:
    - Count per Department
    - Mean Salary per Department
    - Top scorer per Department
13. Create a pivot table showing mean Salary by Dept × City.
```

````python
# 11
employees = [
    {
        "Name": "sanket",
        "Department": "HR",
        "City": "Pune",
        "Salary": 40000,
        "Score": 75
    },
    {
        "Name": "Priya",
        "Department": "DS",
        "City": "Mumbai",
        "Salary": 70000,
        "Score": 92
    },
    {
        "Name": "swetali",
        "Department": "IT",
        "City": "Pune",
        "Salary": 50000,
        "Score": 85
    },
    {
        "Name": "rajnandi",
        "Department": "DS",
        "City": "Mumbai",
        "Salary": 80000,
        "Score": 95
    },
    {
        "Name": "pranu",
        "Department": "HR",
        "City": "Delhi",
        "Salary": 45000,
        "Score": 82
    },
    {
        "Name": "suraj",
        "Department": "IT",
        "City": "Delhi",
        "Salary": 60000,
        "Score": 78
    }
]
df = pd.DataFrame(employees)
print("Employee Data")
print(df)
result = df[df["Score"] > 80]
result = result.sort_values(by="Score", ascending=False)

print("Employees with Score > 80")
print(result)
````

````python
# 12
# 1
count = df.groupby("Department")["Name"].count()

print("Employee Count")
print(count)
# 2
average_salary = df.groupby("Department")["Salary"].mean()

print("Average Salary")
print(average_salary)
#3
top_scorer = df.groupby("Department")["Score"].max()
print("Top Score in Each Department")
print(top_scorer)
````

````python
# 13
df = pd.DataFrame(employees)
print(df)

pivot_table = df.pivot_table(
    values="Salary",        # Column to calculate
    index="Department",     # Rows
    columns="City",         # Columns
    aggfunc="mean"          # Average Salary
)
print(pivot_table)
````

### Exercise Set 5 — DateTime & I/O
```
14. Parse a list of date strings, extract year/month/weekday.
15. Calculate the number of days between two date columns.
16. Read a CSV, optimize dtypes, compare memory before/after.
```

````python
df = pd.DataFrame({
    "Joining_Date": [
        "2024-01-15",
        "2023-05-20",
        "2022-10-08",
        "2021-12-25"
    ]
})

print("Original DataFrame")
print(df)
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])
df["Year"] = df["Joining_Date"].dt.year
df["Month"] = df["Joining_Date"].dt.month
df["Weekday"] = df["Joining_Date"].dt.day_name()

print("After Extracting Date Information")
print(df)
````

````python
# 15
df = pd.DataFrame({
    "Start_Date": [
        "2024-01-01",
        "2024-02-10",
        "2024-03-15"
    ],
    "End_Date": [
        "2024-01-20",
        "2024-02-25",
        "2024-04-10"
    ]
})

print("Original DataFrame")
print(df)
df["Start_Date"] = pd.to_datetime(df["Start_Date"])
df["End_Date"] = pd.to_datetime(df["End_Date"])
df["Days"] = (df["End_Date"] - df["Start_Date"]).dt.days
print(df)
````

````python
import pandas as pd

# Create a dummy employees.csv file for demonstration
# This data is consistent with the 'employees' list used in previous exercises
csv_data = '''Name,Department,City,Salary,Score
sanket,HR,Pune,40000,75
Priya,DS,Mumbai,70000,92
swetali,IT,Pune,50000,85
rajnandi,DS,Mumbai,80000,95
pranu,HR,Delhi,45000,82
suraj,IT,Delhi,60000,78
'''
with open('employees.csv', 'w') as f:
    f.write(csv_data)

# Step 1: Read the CSV file
df = pd.read_csv("employees.csv")
print("Original DataFrame")
print(df)
# Step 2: Check memory usage

print("Memory Usage Before Optimization")
print(df.memory_usage(deep=True))
# Step 3: Change data types
# Convert text columns to category
# This saves memory when values repeat.
df["Name"] = df["Name"].astype("category")
df["Department"] = df["Department"].astype("category")
df["City"] = df["City"].astype("category")

# Convert Salary from int64 to int32
# int32 uses less memory.
df["Salary"] = df["Salary"].astype("int32")

# Convert Score from int64 to int8
df["Score"] = df["Score"].astype("int8")
print("Optimized DataFrame")
print(df)

# Step 5: Check memory again
print("Memory Usage After Optimization")
print(df.memory_usage(deep=True))
# Step 6: Calculate total memory

# Read the CSV again to get the original memory usage
original_df = pd.read_csv("employees.csv")

# Total memory before optimization
memory_before = original_df.memory_usage(deep=True).sum()

# Total memory after optimization
memory_after = df.memory_usage(deep=True).sum()
# Step 7: Print results
print("Total Memory Before Optimization :", memory_before, "bytes")
print("Total Memory After Optimization  :", memory_after, "bytes")

# Find memory saved
memory_saved = memory_before - memory_after

print("Memory Saved                     :", memory_saved, "bytes")
````
