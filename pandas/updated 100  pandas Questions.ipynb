### Q56. Detect outliers in a column.
> 💡 **Hint:** Use the **IQR method**: compute Q1, Q3, IQR = Q3–Q1. Outliers are values below `Q1 - 1.5*IQR` or above `Q3 + 1.5*IQR`. Alternatively use Z-score: `|z| > 3`.

```python
df3 = pd.DataFrame({
    'id': [1, 2, 3, 4, 5],
    'prices': [200, 400, 350, 20000,10]
})
df3
```

```python
q1 = df3['prices'].quantile(0.25)
q3 = df3['prices'].quantile(0.75)
iqr = q3-q1
iqr

low_limit = q1 - 1.5*iqr
up_limit = q3 + 1.5*iqr

df3[(df3['prices']> up_limit) |(df3['prices']< low_limit)]
```

### Q100. Create an end-to-end analysis workflow.
> 💡 **Hint:** Follow this pipeline:  
> 1. Load data → 2. Inspect (shape, dtypes, nulls) → 3. Clean (fillna, drop_duplicates) → 4. Feature engineer → 5. GroupBy/pivot analysis → 6. Sort and rank → 7. Export results.

```python
raw_data = {
    'Employee': ['Alice', 'Bob', 'Charlie', 'Alice', 'David', 'Eva', 'Bob'],
    'Department': ['HR', 'IT', 'IT', 'HR', 'Sales', 'Sales', 'IT'],
    'Salary': [50000, 60000, np.nan, 50000, 70000, 80000, 60000],
    'Performance_Score': [4.2, 3.8, 4.5, 4.2, 2.9, 4.7, 3.8]
}

df6 = pd.DataFrame(raw_data)
```

```python
df6
```

```python
#inspect

print("--- 1. Shape ---")
print(df6.shape)  # Output: (7, 4)

print("\n--- 2. Data Types ---")
print(df6.dtypes)

print("\n--- 3. Missing Values ---")
print(df6.isnull().sum())  # Shows 'Salary' has 1 missing value
```

```python
#Clean
# Drop duplicates (removes the duplicate Alice and Bob entries)
df6 = df6.drop_duplicates()

# Fill missing Salary with the column mean
mean_salary = df6['Salary'].mean()
df6['Salary'] = df6['Salary'].fillna(mean_salary)
```

```python
#Feature Engineer
# Bonus = 10% of salary if performance is above 4.0, otherwise 5%
df6['Bonus'] = np.where(df6['Performance_Score'] > 4.0, df6['Salary'] * 0.10, df6['Salary'] * 0.05)
```

```python
#GroupBy / Pivot Analysis
summary_df = df6.groupby('Department')[['Salary', 'Bonus']].sum()
```

```python
#sort and Rank

# Sort descending
final_report = summary_df.sort_values(by='Bonus', ascending=False)
```

```python
#export Result
final_report.to_csv('department_bonus_report.csv')
print("Workflow complete! File saved to your project directory.")
```
