#### 62. Considering two arrays with shape (1,3) and (3,1), how to compute their sum using an iterator? (★★☆)
`hint: np.nditer`

````python
A = np.array([[1, 2, 3]])

B = np.array([[10],
              [20],
              [30]])


result = np.zeros((3, 3))


for i, a in np.ndenumerate(A):
    for j, b in np.ndenumerate(B):
        # Corrected indexing: j[0] for row and i[1] for column to achieve broadcasted sum
        result[j[0], i[1]] = a + b

print(result)
````

#### 74. Given a sorted array C that corresponds to a bincount, how to produce an array A such that np.bincount(A) == C? (★★★)
`hint: np.repeat`

````python
C = np.array([0, 2, 1, 0, 3])

# Generate a sequence of indices from 0 to len(C)-1
indices = np.arange(len(C))

# Repeat each index according to the counts in C
A = np.repeat(indices, C)

print("Reconstructed array A:", A)
print("Is np.bincount(A) == C?", np.array_equal(np.bincount(A), C))
````

#### 75. How to compute averages using a sliding window over an array? (★★★)

#### 78. Consider 2 sets of points P0,P1 describing lines (2d) and a point p, how to compute distance from p to each line i (P0[i],P1[i])? (★★★)
`No hints provided...`

````python
def distance_to_lines(P0, P1, p):
    # Vector representing each line segment (direction)
    U = P1 - P0  # Shape: (N, 2)
    
    # Square of the length of each line segment
    L = (U**2).sum(axis=1)  # Shape: (N,)
    
    # Projection factor, clamped between 0 and 1 to stay on the segments
    # formula: projection = (v dot u) / |u|^2
    # We use axis=1 to perform the dot product row-by-row
    projection = ((p - P0) * U).sum(axis=1) / L
    projection = np.clip(projection, 0, 1)  # Shape: (N,)
    
    # Find the closest point on each line segment to point p
    closest_points = P0 + projection[:, np.newaxis] * U  # Shape: (N, 2)
    
    # Compute the Euclidean distance from p to the closest points
    distances = np.sqrt(((closest_points - p)**2).sum(axis=1))  # Shape: (N,)
    
    return distances

# --- Example Usage ---
# 3 lines in 2D space
P0 = np.array([[0, 0], [1, 2], [5, 5]])
P1 = np.array([[0, 4], [4, 2], [8, 5]])

# The point we want to measure from
p = np.array([2, 2])

print("Distances to each line segment:", distance_to_lines(P0, P1, p))
````

#### 79. Consider 2 sets of points P0,P1 describing lines (2d) and a set of points P, how to compute distance from each point j (P[j]) to each line i (P0[i],P1[i])? (★★★)
`No hints provided...`

````python
def point_to_line_distance_matrix(P0, P1, P):
    # U: Line direction vectors. Shape: (num_lines, 2)
    U = P1 - P0
    
    # L: Squared lengths of each line segment. Shape: (num_lines,)
    L = (U**2).sum(axis=1)
    
    # Using broadcasting to find vectors from P0 to ALL points in P
    # P[:, np.newaxis, :] has shape (num_points, 1, 2)
    # P0[np.newaxis, :, :] has shape (1, num_lines, 2)
    # Vector V shape: (num_points, num_lines, 2)
    V = P[:, np.newaxis, :] - P0[np.newaxis, :, :]
    
    # Compute dot products between V and U for all combinations
    # U[np.newaxis, :, :] has shape (1, num_lines, 2)
    # We sum along axis=2 (the X and Y coordinates)
    dot_products = (V * U[np.newaxis, :, :]).sum(axis=2) # Shape: (num_points, num_lines)
    
    # Calculate projection factors and clamp between 0 and 1
    # L[np.newaxis, :] allows broadcasting across all points
    projection = dot_products / L[np.newaxis, :]
    projection = np.clip(projection, 0, 1) # Shape: (num_points, num_lines)
    
    # Find the closest point on every line segment for every point in P
    # projection[:, :, np.newaxis] has shape (num_points, num_lines, 1)
    # U[np.newaxis, :, :] has shape (1, num_lines, 2)
    closest_points = P0[np.newaxis, :, :] + projection[:, :, np.newaxis] * U[np.newaxis, :, :]
    
    # Compute the Euclidean distance from each point in P to its closest points
    # Diff shape: (num_points, num_lines, 2)
    diff = closest_points - P[:, np.newaxis, :]
    distances = np.sqrt((diff**2).sum(axis=2)) # Shape: (num_points, num_lines)
    
    return distances

# --- Example Usage ---
P0 = np.array([[0, 0], [1, 2]])  # 2 lines
P1 = np.array([[0, 4], [4, 2]])

P = np.array([[2, 2], [0, 2], [5, 5]])  # 3 points

matrix = point_to_line_distance_matrix(P0, P1, P)
print("Distance Matrix (Points x Lines):\n", matrix)
````

#### 80. Consider an arbitrary array, write a function that extracts a subpart with a fixed shape and centered on a given element (pad with a `fill` value when necessary) (★★★)
`hint: minimum maximum`

````python

def extract_subpart(Z, shape, center, fill=0):
    # 1. Initialize the output array completely filled with the pad value
    output = np.full(shape, fill_value=fill, dtype=Z.dtype)
    
    # Convert inputs to numpy arrays for element-wise math
    shape = np.array(shape)
    center = np.array(center)
    
    # 2. Calculate bounds relative to the center
    # S // 2 gives the radius of the window on each side
    half_shape = shape // 2
    
    # Start and end coordinates inside the source array Z
    z_start = center - half_shape
    z_end = z_start + shape
    
    # 3. Clamp the source coordinates so they stay within Z's actual boundaries
    # This is where we use the minimum and maximum hint
    z_start_clamped = np.maximum(z_start, 0)
    z_end_clamped = np.minimum(z_end, Z.shape)
    
    # 4. Calculate where to paste this valid slice inside our output array
    out_start = z_start_clamped - z_start
    out_end = out_start + (z_end_clamped - z_start_clamped)
    
    # 5. Extract from Z and paste into output if there's a valid overlap
    if np.all(z_start_clamped < z_end_clamped):
        # Create slice objects dynamically for multi-dimensional arrays
        z_slices = tuple(slice(s, e) for s, e in zip(z_start_clamped, z_end_clamped))
        out_slices = tuple(slice(s, e) for s, e in zip(out_start, out_end))
        
        output[out_slices] = Z[z_slices]
        
    return output

# --- Example Usage ---
# Create a 6x6 array of counting numbers
Z = np.arange(1, 37).reshape(6, 6)
print("Original Array Z:\n", Z)

print("\n" + "="*40 + "\n")

# Case A: Perfectly centered inside the array
print("Centered at (3,3), Shape (3,3):")
print(extract_subpart(Z, shape=(3, 3), center=(3, 3), fill=-1))

print("\n" + "="*40 + "\n")

# Case B: Near the top-left corner (requires padding)
print("Centered at (0,0), Shape (3,3) - Overflows top and left borders:")
print(extract_subpart(Z, shape=(3, 3), center=(0, 0), fill=-1))
````

#### 85. Create a 2D array subclass such that Z[i,j] == Z[j,i] (★★★)
`hint: class method`

````python
class SymmetricArray(np.ndarray):
    def __setitem__(self, index, value):
        # Handle 2D index unpacking (e.g., Z[i, j])
        if isinstance(index, tuple) and len(index) == 2:
            i, j = index
            # Set the requested element (i, j)
            super().__setitem__((i, j), value)
            # Mirror the assignment to the symmetric element (j, i)
            super().__setitem__((j, i), value)
        else:
            # Fallback for alternative indexing styles
            super().__setitem__(index, value)

def make_symmetric(array):
    """
    Helper function to convert a standard array into a SymmetricArray subclass.
    It forces the input matrix to be mathematically symmetric first.
    """
    arr = np.asarray(array)
    # Force mathematical symmetry: (A + A.T) / 2 (ignoring diagonal doubling)
    symmetric_data = arr + arr.T - np.diag(arr.diagonal())
    return symmetric_data.view(SymmetricArray)

# --- Example Usage ---
# Create a 4x4 instance
Z = make_symmetric(np.zeros((4, 4), dtype=int))

print("Initial Matrix:")
print(Z)

print("\nAssigning Z[0, 2] = 42...")
Z[0, 2] = 42

print("\nUpdated Matrix (Notice both Z[0,2] and Z[2,0] changed):")
print(Z)

print(f"\nIs Z[0, 2] == Z[2, 0]? {Z[0, 2] == Z[2, 0]}")
````

#### 88. How to implement the Game of Life using numpy arrays? (★★★)
`No hints provided...`

````python
def game_of_life_step(Z):
    """
    Computes the next generation of the Game of Life grid.
    Z is a 2D array where 1 represents a live cell and 0 is a dead cell.
    """
    # Create an array to count neighbors, ignoring the outermost border cells
    # N will hold the sum of the 8 neighbors for every cell from index 1 to -2
    N = (Z[0:-2, 0:-2] + Z[0:-2, 1:-1] + Z[0:-2, 2:] +
         Z[1:-1, 0:-2]                 + Z[1:-1, 2:] +
         Z[2:  , 0:-2] + Z[2:  , 1:-1] + Z[2:  , 2:])

    # Apply Conway's rules to the internal grid:
    # 1. A dead cell with exactly 3 neighbors becomes a live cell (Birth)
    birth = (N == 3) & (Z[1:-1, 1:-1] == 0)
    
    # 2. A live cell with 2 or 3 neighbors stays alive (Survival)
    survive = ((N == 2) | (N == 3)) & (Z[1:-1, 1:-1] == 1)

    # Reset the entire grid to dead (in-place)
    Z[...] = 0
    
    # Set cells that match birth or survive conditions back to live (1)
    Z[1:-1, 1:-1][birth | survive] = 1
    
    return Z

# --- Example Usage ---
# Initialize a random 6x6 grid with 0s and 1s
np.random.seed(42)  # For reproducible results
grid = np.random.randint(0, 2, (6, 6))

print("Generation 0:\n", grid)

# Run one step
grid = game_of_life_step(grid)

print("\nGeneration 1:\n", grid)
````

#### 90. Given an arbitrary number of vectors, build the cartesian product (every combination of every item) (★★★)
`hint: np.indices`

````python
def cartesian_product(vectors):
    # Convert all inputs to standard numpy arrays
    vectors = [np.asarray(v) for v in vectors]
    
    # Get the length of each vector to define the grid dimensions
    shape = [len(v) for v in vectors]
    
    # 1. Generate an index mesh grid based on the dimensions of the vectors
    # This creates an array of index positions for every combination
    idx = np.indices(shape, dtype=int)
    
    # 2. Use advanced indexing to map the index positions back to the actual values
    # Then stack them vertically and reshape into a 2D matrix of combinations
    num_vectors = len(vectors)
    product = np.vstack([vectors[i][idx[i]] for i in range(num_vectors)])
    
    # Reshape to have one combination per row
    return product.reshape(num_vectors, -1).T

# --- Example Usage ---
v1 = [1, 2, 3]
v2 = [4, 5]
v3 = [6, 7]

result = cartesian_product((v1, v2, v3))
print("Total number of combinations:", len(result))
print("\nAll combinations (Cartesian Product):\n", result)
````

#### 98. Considering a path described by two vectors (X,Y), how to sample it using equidistant samples (★★★)?
`hint: np.cumsum, np.interp`

````python
def sample_equidistant_path(X, Y, num_samples=10):
    # 1. Compute the differences between consecutive coordinates
    dx = np.diff(X)
    dy = np.diff(Y)
    
    # 2. Calculate the Euclidean distance of each individual segment
    segment_distances = np.sqrt(dx**2 + dy**2)
    
    # 3. Build a running total of the distance along the path
    # We prepend a 0 so that the distance profile matches the length of X and Y
    cumulative_distance = np.concatenate(([0], np.cumsum(segment_distances)))
    total_length = cumulative_distance[-1]
    
    # 4. Generate a new grid of completely equidistant distances
    equidistant_grid = np.linspace(0, total_length, num_samples)
    
    # 5. Interpolate X and Y coordinates onto the new uniform distance grid
    X_samples = np.interp(equidistant_grid, cumulative_distance, X)
    Y_samples = np.interp(equidistant_grid, cumulative_distance, Y)
    
    return X_samples, Y_samples

# --- Example Usage ---
# A path with highly irregular segments
X = np.array([0.0, 1.0, 1.5, 4.0, 5.0])
Y = np.array([0.0, 0.5, 2.0, 2.2, 4.0])

# Resample it into 6 perfectly equidistant points along the line curve
X_new, Y_new = sample_equidistant_path(X, Y, num_samples=6)

print("Original X:", X)
print("Resampled X (Equidistant):", np.round(X_new, 2))
print("\nOriginal Y:", Y)
print("Resampled Y (Equidistant):", np.round(Y_new, 2))

# Verification: Calculate the segment lengths of the new resampled points
new_dx = np.diff(X_new)
new_dy = np.diff(Y_new)
new_segments = np.sqrt(new_dx**2 + new_dy**2)
print("\nVerifying new segment lengths (should all be equal):", np.round(new_segments, 4))
````

#### 100. Compute bootstrapped 95% confidence intervals for the mean of a 1D array X (i.e., resample the elements of an array with replacement N times, compute the mean of each sample, and then compute percentiles over the means). (★★★)
`hint: np.percentile`

````python
# 1. Create a sample 1D dataset (e.g., 100 elements)
np.random.seed(42)  # For reproducible results
X = np.random.normal(loc=10.0, scale=2.0, size=100)

# 2. Define the number of bootstrap resamples (typically 1,000 to 10,000)
N = 10000

# 3. Resample the elements of X with replacement N times at once
# This creates a 2D array of shape (N, len(X))
bootstrap_samples = np.random.choice(X, size=(N, len(X)), replace=True)

# 4. Compute the mean of each individual sample row (axis=1)
bootstrap_means = bootstrap_samples.mean(axis=1)

# 5. Compute the 2.5th and 97.5th percentiles to get the 95% confidence interval
confidence_interval = np.percentile(bootstrap_means, [2.5, 97.5])

print(f"Dataset Observed Mean: {X.mean():.4f}")
print(f"Bootstrapped 95% Confidence Interval: [{confidence_interval[0]:.4f}, {confidence_interval[1]:.4f}]")
````
