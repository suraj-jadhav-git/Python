# Series

A Series is a one-dimensional labelled array capable of holding heterogeneous data.

## Components
- Data
- Index
- Data Type

## Facts
- Mutable values
- Immutable size
- Supports vectorized operations

```python
# Create Series from list
s1 = pd.Series([10,20,30], index=['a','b','c'])
print(s1)

# Create Series from dictionary
s2 = pd.Series({'Math':90,'Science':95,'English':88})
print(s2)
```

# DataFrame

A DataFrame is a two-dimensional labelled data structure.

## Properties
1. Heterogeneous Data
2. Size Mutable
3. Labelled Axes
4. Supports Missing Values
5. Powerful Indexing

```python
# List of lists
df1 = pd.DataFrame(
    [['John',25],['Mary',30]],
    columns=['Name','Age']
)
print(df1)

# List of dictionaries
df2 = pd.DataFrame([
    {'Name':'Alice','Age':28},
    {'Name':'Bob','Age':35}
])
print(df2)

# Dictionary of Series
df3 = pd.DataFrame({
    'A': pd.Series([1,2,3]),
    'B': pd.Series([4,5,6])
})
print(df3)
```

# Adding Columns

Methods:
- Assign scalar value
- Assign list
- Assign Series

```python
df = pd.DataFrame({
    'Name':['John','Mary','Alice'],
    'Age':[25,30,28]
})

# Scalar
df['Country'] = 'India'

# List
df['Score'] = [80,90,85]

# Series
bonus = pd.Series([5,10,15])
df['Bonus'] = bonus

print(df)
```

# Rename Columns

rename(columns={}, inplace=False)

Parameters:
- columns
- inplace

```python
# Rename single column
df.rename(columns={'Age':'Employee_Age'}, inplace=True)

# Rename multiple columns
df.rename(columns={
    'Score':'Exam_Score',
    'Bonus':'Extra_Bonus'
}, inplace=True)

print(df)
```

# Indexing and Data Selection

## Types
- loc (label based)
- iloc (position based)

```python
employees = pd.DataFrame({
    'Name':['John','Mary','Alice','Bob'],
    'Department':['IT','HR','IT','Finance'],
    'Salary':[50000,60000,55000,70000]
}, index=['E1','E2','E3','E4'])

print(employees)
```

## loc Examples

```python
# Access row
print(employees.loc['E1'])

# Access column
print(employees.loc[:, 'Name'])

# Multiple rows and columns
print(employees.loc[['E1','E3'], ['Name','Salary']])

# Conditional indexing
print(employees.loc[employees['Salary'] > 55000])
```

## iloc Examples

```python
# Single row
print(employees.iloc[0])

# Single column
print(employees.iloc[:,0])

# Subset
print(employees.iloc[0:3, 0:2])
```

1. Difference between Series and DataFrame?

A Series is a one-dimensional array-like object representing a single column of data with an index. A DataFrame is a two-dimensional, size-mutable, tabular data structure with labeled rows and columns. You can think of a DataFrame as a dictionary or collection of multiple Series objects that all share the same index.

2. loc vs iloc?

loc is strictly label-based data selection, meaning you filter data using the actual row and column names (like 'Date' or 'Employee_A'). iloc is integer position-based selection, meaning you select data based on its physical location or index number (like 0, 1, or 2), regardless of what the labels are.

3. Why is Pandas built on NumPy?

Pandas relies on NumPy's ndarray object under the hood, which is written in C and highly optimized for speed and memory efficiency. By building on NumPy, Pandas inherits the ability to perform fast, vectorized mathematical operations across large datasets. It essentially takes NumPy's computational power and adds higher-level data manipulation tools and labels.

4. What is heterogeneous data?

Heterogeneous data refers to tabular data where different columns hold completely different data types. For example, a single dataset might have a 'Name' column (strings), an 'Age' column (integers), and a 'Salary' column (floats). A Pandas DataFrame handles this seamlessly, unlike standard NumPy arrays which typically require all elements to be the exact same data type.

5. How do you rename columns?

The safest and most common way is using the .rename() method, passing a dictionary mapping old names to new names (e.g., df.rename(columns={'old_name': 'new_name'})). You can also overwrite all column names directly by assigning a new list of strings to the df.columns attribute. The .rename() method is preferred because it allows you to target specific columns without touching the rest.

6. How do you add columns?

You can add a column directly by assigning a list or Series to a new dictionary-style key, like df['New_Column'] = values. If you need to place the column at a specific position, you can use the df.insert() method. Alternatively, the df.assign() method allows you to add columns and return a new DataFrame, which is excellent for method chaining.

7. How does Pandas handle missing values?

Pandas standardizes missing data by using the floating-point value NaN (Not a Number) for numerical/string data, or NaT for datetime data. It provides methods like .isna() to detect these missing spots across your dataset. You can then clean the data by dropping the problematic rows/columns using .dropna() or filling the gaps with default values using .fillna().

8. Explain DataFrame attributes.

Attributes provide metadata and structural details about your DataFrame without modifying the data or requiring parentheses to execute. Common examples include df.shape (returns row and column counts), df.columns (returns column names), and df.dtypes (returns the data type of each column). They essentially describe what the DataFrame is, rather than performing an action on it.

9. Difference between list of dictionaries and dictionary of Series?

When creating a DataFrame from a list of dictionaries, Pandas treats each dictionary as a row, with the keys becoming the column headers. When creating it from a dictionary of Series, the keys become the column headers, and each Series is treated as a full column. The dictionary of Series approach also automatically aligns the data based on the indices of each Series.

10. When would you choose loc over iloc?

You should use loc when your DataFrame has meaningful, descriptive index labels (like dates, names, or IDs) and you want to query data based on that specific business logic. loc makes your code highly readable because you are extracting data by its semantic identity rather than an arbitrary physical row number. Additionally, loc slicing is inclusive of the endpoint, which is more intuitive for labeled data.
