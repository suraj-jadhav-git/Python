````python
np.random.seed(0)

X = np.random.rand(200, 5)
y = ((X[:, 0] + X[:, 1]) > 1.0).astype(int)     #If feature0 + feature1 > 1 → class 1 , Otherwise → class 0

# 1. Z-score standardize X column-wise-----------------------------
mean = X.mean(axis=0)           # Column means
std = X.std(axis=0)             # Column stds

X = (X - mean) / std            # Z-score standardize

# 2. Shuffle and split 70/15/15-------------------------------------
shuffle_idx = np.argsort(np.random.rand(200))   #Create a random ordering of row numbers

X = X[shuffle_idx]     #Rearrange X and y in exactly the same random order.
y = y[shuffle_idx]

n = len(X)              # Total samples

n_train = int(0.70 * n)   # Train size
n_val = int(0.15 * n)     # Validation size

X_train = X[:n_train]     # Train features
y_train = y[:n_train]      # # Train labels

X_val = X[n_train:n_train + n_val]      # Validation features
y_val = y[n_train:n_train + n_val]      # Validation labels

X_test = X[n_train + n_val:]            # Test features
y_test = y[n_train + n_val:]            # Test labels

# 3. Initialize weights and bias-----------------------------------------
w = np.zeros(5)
b = 0.0

# 4. Forward pass
z = X_train @ w + b                         # Compute linear score (Xw + b)
y_hat = 1 / (1 + np.exp(-z))                # Convert score to probability using sigmoid

# 5. BCE loss
y_hat_clip = np.clip(y_hat, 1e-7, 1 - 1e-7)     # Avoid log(0) errors

initial_loss = -np.mean(
    y_train * np.log(y_hat_clip)
    + (1 - y_train) * np.log(1 - y_hat_clip)
)                                               # Compute Binary Cross-Entropy (BCE) loss

# 6. Gradient--------------------------------------------------------------
grad_w = X_train.T @ (y_hat - y_train) / len(y_train)       # Compute weight gradients

# 7. One gradient descent step---------------------------------------------
learning_rate = 0.1
w -= learning_rate * grad_w                 # Update weights using gradient descent

# 8. Forward pass again and new BCE loss-----------------------------------
z_new = X_train @ w + b                         # Compute new scores with updated weights
y_hat_new = 1 / (1 + np.exp(-z_new))            # Compute new probabilities

y_hat_new = np.clip(y_hat_new, 1e-7, 1 - 1e-7)          # Avoid log(0) again

updated_loss = -np.mean(
    y_train * np.log(y_hat_new)
    + (1 - y_train) * np.log(1 - y_hat_new)
)                                               # Compute loss after weight update

# 9. Gradient norm-------------------------------------------------------------
grad_norm = np.sqrt((grad_w ** 2).sum())        # Compute gradient magnitude

print("Initial loss :", initial_loss)       # Loss before update
print("Updated loss :", updated_loss)        # Loss after update
print("Gradient norm:", grad_norm)          # Gradient size
````
